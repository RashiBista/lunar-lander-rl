```python
import gymnasium as gym
import numpy as np
from stable_baselines3 import PPO

model = PPO.load("ppo_lunarlander")
env = gym.make("LunarLander-v3")

rewards = []
lengths = []
landings = []

for ep in range(50):
    obs, _ = env.reset(seed=ep)
    total = 0
    steps = 0
    done = False
    terminated = False
    while not done:
        action, _ = model.predict(obs, deterministic=True)
        obs, r, terminated, truncated, _ = env.step(action)
        total += r
        steps += 1
        done = terminated or truncated
    rewards.append(total)
    lengths.append(steps)
    landings.append(terminated and total > 0)

env.close()

print(f"Mean reward:  {np.mean(rewards):.1f} ± {np.std(rewards):.1f}")
print(f"Min / Max:    {np.min(rewards):.1f} / {np.max(rewards):.1f}")
print(f"Landing rate: {np.mean(landings)*100:.1f}%")
print(f"Mean length:  {np.mean(lengths):.0f}")
```

```python
import gymnasium as gym
import numpy as np
from stable_baselines3 import PPO

model = PPO.load("ppo_lunarlander")
env = gym.make("LunarLander-v3")

rewards, lengths, landings = [], [], []

for ep in range(50):
    obs, _ = env.reset(seed=ep)
    total, steps, done, terminated = 0, 0, False, False
    while not done:
        action, _ = model.predict(obs, deterministic=True)
        obs, r, terminated, truncated, _ = env.step(action)
        total += r
        steps += 1
        done = terminated or truncated
    rewards.append(total)
    lengths.append(steps)
    landings.append(terminated and total > 0)

env.close()

print(f"Mean reward:  {np.mean(rewards):.1f} ± {np.std(rewards):.1f}")
print(f"Min / Max:    {np.min(rewards):.1f} / {np.max(rewards):.1f}")
print(f"Landing rate: {np.mean(landings)*100:.1f}%")
print(f"Mean length:  {np.mean(lengths):.0f}")
```
